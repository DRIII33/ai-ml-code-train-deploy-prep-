## Week 9 — Project 2: Exploratory Data Analysis and Long-Tail Discovery

**Syllabus focus:** Project 2 — exploratory data analysis and visualization; EDA on a real-world dataset; discovery of patterns and anomalies.

**Business context:** We now know the dataset is cleaned. The next step is to look for patterns that reveal whether the discovery pipeline favors mainstream artists over emerging ones.

**Capstone deliverable:** An EDA report showing where bias and under-discovery exist in the music ecosystem.

### Learning objectives
- summarize artist and genre performance
- identify outliers and unusual patterns
- compare mainstream vs long-tail exposure
- write a data-story using analytical evidence


```python
import pandas as pd

df = pd.DataFrame({
    'artist_name': ['Nova Vela', 'Luna Harbor', 'Aster Vale', 'Bright Echo', 'Velvet Foundry', 'Sunset Thread', 'Nimbus Cove'],
    'genre': ['indie pop', 'alternative rock', 'ambient electronic', 'hip-hop', 'indie pop', 'electronic', 'jazz'],
    'monthly_listeners': [42000, 26000, 7100, 54000, 15000, 6300, 3100],
    'playlist_inclusion': [1, 1, 0, 1, 0, 0, 0],
    'new_artist': [0, 0, 1, 0, 1, 1, 1],
    'streams_last_30d': [120000, 98000, 31000, 152000, 24000, 20500, 7000]
})

print(df.describe())
print('
Artist exposure by playlist inclusion:
')
print(df.groupby('playlist_inclusion')['streams_last_30d'].mean())
```

## EDA questions to answer
- Which artist groups are the highest performers?
- Which genres appear underrepresented in playlists?
- Are new artists limited by low exposure?
- Where do the strongest opportunities for discovery appear?


```python
summary = df.groupby('genre').agg(
    avg_listeners=('monthly_listeners', 'mean'),
    total_streams=('streams_last_30d', 'sum'),
    artist_count=('artist_name', 'count')
)

print(summary)
```

## Interpretation
If `hip-hop` and `indie pop` dominate streams while `jazz` or `ambient electronic` appear with low playlist inclusion, that indicates a structural gap in discovery. This observation is central to the fairness problem in the final capstone.
