## Why this matters in the music industry

Recommendation systems are not neutral. When they are trained on historical engagement, they can systematically reward already-popular artists and under-recognize niche music. In practice, this creates a feedback loop:

**popular artists receive more exposure → more streams → stronger recommendation signal → more reinforcement**

This is a strategic problem for streaming platforms because it affects:
- artist revenue and retention;
- listener discovery and content diversity;
- playlist curation quality;
- long-term brand trust.

# 1. Lesson 1 — Introduction to Data Science in Music

A data-science workflow is simply a structured way to move from a business question to a decision.

**Question → Data → Processing → Analysis → Insight → Decision**

In a music platform context, a business question might be:

> Which artists are under-discovered relative to their engagement potential, and how can we improve discovery without harming user retention?

The relevant data might include track metadata, streams, likes, skip rates, playlists, demographics, and artist identity.


```python
artist_streams = {
    'Nova Vela': 120000,
    'Luna Harbor': 95000,
    'Bright Echo': 78000,
    'Aster Vale': 31000,
    'Velvet Foundry': 29000
}

total_streams = sum(artist_streams.values())
average_streams = total_streams / len(artist_streams)
top_artist = max(artist_streams, key=artist_streams.get)

print('Total streams:', total_streams)
print('Average streams:', average_streams)
print('Top artist:', top_artist)
print('Top artist streams:', artist_streams[top_artist])
```
