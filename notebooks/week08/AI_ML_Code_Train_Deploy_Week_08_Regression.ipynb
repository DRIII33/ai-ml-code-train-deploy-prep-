## Week 8 — Regression and Popularity Prediction

**Focus:** build a baseline model to predict artist or track popularity from observable features.

### Business objective
Estimate whether an artist or track is likely to perform well and whether this can be linked to discovery or fairness issues.

### Deliverable
A regression model and summary of feature importance for the music platform problem.


```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1000, 1800, 2600, 3400, 4200])

model = LinearRegression()
model.fit(X, y)
print('Coefficient:', model.coef_[0])
print('Intercept:', model.intercept_)
print('Prediction for 6:', model.predict([[6]])[0])
```
