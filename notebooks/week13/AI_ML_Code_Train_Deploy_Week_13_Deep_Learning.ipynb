## Example concept
Audio features can be transformed into low-dimensional embeddings. Similar music can then be recommended even when the artists are not well-known.


```python
import numpy as np

tracks = np.array([[0.2, 0.3], [0.21, 0.31], [0.9, 0.8]])
similarity = np.linalg.norm(tracks[0] - tracks[1])
print('Similarity between track 1 and 2:', similarity)
```
