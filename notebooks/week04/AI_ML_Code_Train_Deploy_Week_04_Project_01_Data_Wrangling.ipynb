## Synthetic but realistic dataset

The following data mirrors the problems a platform often sees: duplicate artist names, mixed-case genres, missing metadata, and inconsistent popularity values.


```python
import pandas as pd

df = pd.DataFrame({
    'artist_id': ['ART-001', 'ART-001', 'ART-002', 'ART-003', 'ART-004', 'ART-005'],
    'artist_name': ['Nova Vela', 'Nova Vela', 'Luna Harbor', 'Aster Vale', 'Bright Echo', None],
    'genre': ['Indie Pop', 'indie pop', 'alt rock', 'Ambient Electronic', 'Hip Hop', 'Ambient Electronic'],
    'monthly_listeners': [42000, 42000, 26000, 7100, 54000, 14000],
    'streams_last_30d': [120000, 120000, 98000, 31000, 152000, 42000],
    'playlist_inclusion': [True, True, True, False, True, False],
    'new_artist': [False, False, False, True, False, True],
    'region': ['US', 'US', 'UK', 'DE', 'US', 'FR']
})

print(df)
print('
Missing values:')
print(df.isnull().sum())
```

## Cleaning workflow

We will:
- remove duplicate records
- fill or drop missing artist names
- normalize genre casing
- create a cleaned dataset for later modeling


```python
clean_df = df.drop_duplicates(subset=['artist_id', 'artist_name', 'genre']).copy()
clean_df['artist_name'] = clean_df['artist_name'].fillna('Unknown artist')
clean_df['genre'] = clean_df['genre'].str.strip().str.lower()
clean_df['genre'] = clean_df['genre'].replace({'alt rock': 'alternative rock', 'hip hop': 'hip-hop'})

print(clean_df)
print('
Final row count:', len(clean_df))
```

## Quality report

A good data quality report should capture both the issue and the action taken. Example:
- duplicate rows were removed
- missing artist names were filled with a placeholder
- genre labels normalized to lowercase standards
- data is now ready for analysis


```python
quality_report = {
    'initial_rows': len(df),
    'final_rows': len(clean_df),
    'duplicates_removed': len(df) - len(clean_df),
    'missing_artist_names': int(df['artist_name'].isnull().sum()),
    'genre_standardized': True
}

print(quality_report)
```
