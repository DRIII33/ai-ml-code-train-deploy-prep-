## Example: artist stream distribution

Suppose we have stream counts for a set of artists. A few top artists may receive the majority of streams. This is not automatically bad, but it matters if the distribution is too concentrated and leaves consistent discovery opportunities for long-tail artists.


```python
stream_counts = [120000, 98000, 31000, 45000, 14000, 9000, 6500, 3500, 2800, 1200]

mean_streams = sum(stream_counts) / len(stream_counts)
median_streams = sorted(stream_counts)[len(stream_counts) // 2]
min_streams = min(stream_counts)
max_streams = max(stream_counts)

print('Mean streams:', mean_streams)
print('Median streams:', median_streams)
print('Min streams:', min_streams)
print('Max streams:', max_streams)
```

```python
import statistics

stdev = statistics.pstdev(stream_counts)
variance = statistics.pvariance(stream_counts)

print('Population std dev:', stdev)
print('Population variance:', variance)
```

## Fairness metrics concept

It is useful to compute how concentrated streams are among the top artists. A common question is: what percentage of total streams come from the top 10% of artists?

This helps quantify whether recommendation systems may be overfitting to high-profile talent rather than allowing balanced discovery.


```python
total_streams = sum(stream_counts)
top_3_streams = sum(sorted(stream_counts, reverse=True)[:3])
share_top_3 = top_3_streams / total_streams

print('Total streams:', total_streams)
print('Top 3 share:', share_top_3)
```

## Practice task

Use the following artist metric dictionary and calculate:
- average monthly listeners
- median monthly listeners
- top 20% exposure share
- whether the distribution looks highly concentrated


```python
artist_listeners = {
    'Nova Vela': 42000,
    'Luna Harbor': 26000,
    'Aster Vale': 7100,
    'Bright Echo': 54000,
    'Velvet Foundry': 15000,
    'Sunset Thread': 9000,
    'Nimbus Cove': 6000,
    'Harbor Bloom': 3300
}

values = list(artist_listeners.values())
print('Average listeners:', sum(values) / len(values))
print('Median listeners:', sorted(values)[len(values) // 2])
top_two = sum(sorted(values, reverse=True)[:2])
print('Top 2 share:', top_two / sum(values))
```

## Business interpretation
If a large share of streams comes from a small number of artists, it may indicate that recommendation systems are reinforcing existing popularity rather than promoting equitable discovery. This is precisely the kind of issue to investigate in the capstone project.
