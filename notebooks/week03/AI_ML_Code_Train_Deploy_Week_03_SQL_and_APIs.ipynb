## Core data model

A simplified music platform schema could include the following tables:

- `artists`: artist metadata
- `tracks`: track information
- `listeners`: user demographics and listening profile
- `listening_events`: track listens, skips, likes
- `playlists`: curated playlists and placements
- `recommendations`: model outputs and ranking scores


````python
artists = [
    {'artist_id': 'ART-001', 'artist_name': 'Nova Vela', 'genre': 'indie pop', 'monthly_listeners': 42000},
    {'artist_id': 'ART-002', 'artist_name': 'Luna Harbor', 'genre': 'alt rock', 'monthly_listeners': 26000},
    {'artist_id': 'ART-003', 'artist_name': 'Aster Vale', 'genre': 'ambient electronic', 'monthly_listeners': 7100},
    {'artist_id': 'ART-004', 'artist_name': 'Bright Echo', 'genre': 'hip hop', 'monthly_listeners': 54000},
    {'artist_id': 'ART-005', 'artist_name': 'Velvet Foundry', 'genre': 'indie pop', 'monthly_listeners': 15000}
]

print('Artists loaded:', len(artists))
for artist in artists:
    print(artist['artist_name'], artist['monthly_listeners'])
````

## Database design in plain SQL

Example SQL schema:

```sql
CREATE TABLE artists (
    artist_id TEXT PRIMARY KEY,
    artist_name TEXT NOT NULL,
    genre TEXT,
    monthly_listeners INTEGER
);

CREATE TABLE listening_events (
    event_id TEXT PRIMARY KEY,
    listener_id TEXT,
    artist_id TEXT,
    track_id TEXT,
    event_type TEXT,
    event_timestamp TIMESTAMP
);
```

These tables allow the business team to answer questions such as:
- Which artists are underrepresented in playlists?
- Which genres receive the greatest share of engagement?
- Which listeners are most likely to try new artists?


## SQL practice questions

Write SQL statements that answer the following:
1. List all artists with more than 30,000 monthly listeners.
2. Count how many artists are in each genre.
3. Find the average monthly listeners by genre.
4. Rank artists by monthly listeners descending.


````python
import pandas as pd

artist_df = pd.DataFrame(artists)

high_listener_artists = artist_df[artist_df['monthly_listeners'] > 30000]
print('High-listener artists:\n', high_listener_artists[['artist_name', 'monthly_listeners']])

genre_counts = artist_df.groupby('genre').size().reset_index(name='artist_count')
print('\nGenre counts:\n', genre_counts)

genre_avg = artist_df.groupby('genre')['monthly_listeners'].mean().reset_index(name='avg_listeners')
print('\nAverage listeners by genre:\n', genre_avg)

ranked = artist_df.sort_values('monthly_listeners', ascending=False)
print('\nRanked artists:\n', ranked[['artist_name', 'monthly_listeners']])
````
