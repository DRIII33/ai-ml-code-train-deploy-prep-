## Week 7 — Data Visualization and Discovery Reporting

**Syllabus focus:** data visualization with Matplotlib and Seaborn; basic and advanced plots; customizations.

**Business context:** Recommendation systems are easier to trust when stakeholders can see them. Visualizations help explain whether long-tail artists are gaining traction or whether a platform is overfitting to the most popular creators.

**Capstone deliverable:** A dashboard-like visualization summary for music discovery and fairness.

### Learning objectives
- create effective charts for engagement and artist distribution
- communicate business insight through visuals
- compare mainstream and long-tail exposure


```python
import matplotlib.pyplot as plt

artists = ['Nova Vela', 'Bright Echo', 'Luna Harbor', 'Aster Vale', 'Velvet Foundry']
streams = [120000, 152000, 98000, 31000, 24000]

plt.figure(figsize=(10, 5))
plt.bar(artists, streams, color=['#4C72B0', '#55A868', '#C44E52', '#8172B2', '#CCB974'])
plt.title('Artist stream distribution', fontsize=14)
plt.ylabel('Streams')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## Bar chart interpretation

This chart makes it easy to see imbalance. If a few artists dominate the plot, it may be because the recommendation system favors popularity rather than balanced discovery.


```python
labels = ['Mainstream', 'Long-tail']
exposure = [72, 28]

plt.figure(figsize=(6, 6))
plt.pie(exposure, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#4C72B0', '#55A868'])
plt.title('Recommended exposure split')
plt.axis('equal')
plt.show()
```
