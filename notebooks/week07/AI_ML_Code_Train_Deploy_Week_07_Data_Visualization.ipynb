```python
import matplotlib.pyplot as plt

artists = ['Nova Vela', 'Luna Harbor', 'Aster Vale', 'Bright Echo']
streams = [120000, 98000, 31000, 152000]

plt.bar(artists, streams, color=['#4C72B0', '#55A868', '#C44E52', '#8172B2'])
plt.title('Artist stream distribution')
plt.ylabel('Streams')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
